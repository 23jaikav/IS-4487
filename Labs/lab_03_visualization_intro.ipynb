
## 4. Create basic charts

Pandas makes it easy to create data visualizations directly from a DataFrame using built-in charting functions. With simple commands like df.plot(), you can quickly generate line charts, bar graphs, histograms, and more without needing a lot of extra code. These visualizations help you see patterns, trends, and relationships in your data that might not be obvious from raw numbers alone. By learning how to use pandas’ plotting tools, you can turn your analysis into clear, visual insights that are easier to understand and communicate.

Below are examples of the following chart type:
- Line chart
- Bar chart
- Histogram
- Box plot
- Scatter plot




```python
#Line chart
df.plot(x='carmodel', y='mpg', kind='line', title="MPG by Car")
```

```python
#Bar chart
df.plot(x='carmodel', y='mpg', kind='bar', title="MPG by Car")
```

```python
#Histogram
df['mpg'].plot(kind='hist', bins=10, title="MPG Distribution")
```

```python
#Boxplot
df['mpg'].plot(kind='box', title = 'Miles Per Gallon')
```

```python
#Scatter plot
df.plot.scatter(x='mpg',y='hp',c='wt')
```

- 2 Create a bar chart of cars and HP, with different colors for each car model (you are allowed to import another library, other than Pandas to do this)
- 3 Create a scatter plot of wt (weight) vs. mpg.

```python
import matplotlib.pyplot as plt

transmission_counts = df['am'].value_counts()

# Create a pie chart
fig = transmission_counts.plot.pie(autopct='%1.1f%%', figsize=(6, 6), labels=['Automatic', 'Manual'], title='Transmission Type Distribution')

# The 'labels' parameter in plot.pie handles the labels for the wedges directly.
# If we wanted to adjust the text properties, we might iterate over the text objects, but for simple labels, the above is sufficient.

plt.ylabel('') # Hide the default 'am' label on the y-axis for a cleaner look
plt.show()
```

```python
import matplotlib.pyplot as plt

# Generate a list of colors, one for each car model
num_cars = len(df['carmodel'].unique())
colors = plt.cm.tab20(range(num_cars)) # Using a colormap for distinct colors

# Create a bar chart of HP by Car model with different colors for each car
fig, ax = plt.subplots(figsize=(15, 7)) # Adjust figure size for better readability
df.plot(x='carmodel', y='hp', kind='bar', ax=ax, color=colors, title='Horsepower by Car Model')

plt.xlabel('Car Model')
plt.ylabel('Horsepower (HP)')
plt.xticks(rotation=90) # Rotate x-axis labels for better readability
plt.tight_layout() # Adjust layout to prevent labels from being cut off
plt.show()
```

```python
import matplotlib.pyplot as plt

# Create a scatter plot of wt (weight) vs. mpg
fig, ax = plt.subplots(figsize=(8, 6))
df.plot.scatter(x='wt', y='mpg', ax=ax, title='Weight vs. Miles Per Gallon')

plt.xlabel('Weight (1000 lbs)')
plt.ylabel('Miles Per Gallon (MPG)')
plt.grid(True, linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Calculate the average mpg for each cylinder type
average_mpg_by_cyl = df.groupby('cyl')['mpg'].mean()

# Create a bar chart
fig, ax = plt.subplots(figsize=(8, 6))
average_mpg_by_cyl.plot(kind='bar', ax=ax, title='Average MPG by Number of Cylinders')

plt.xlabel('Number of Cylinders')
plt.ylabel('Average Miles Per Gallon (MPG)')
plt.xticks(rotation=0) # No rotation needed for few categories
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

# Plot a histogram of the weight column
fig, ax = plt.subplots(figsize=(8, 6))
df['wt'].plot(kind='hist', bins=10, ax=ax, title='Distribution of Car Weight')

plt.xlabel('Weight (1000 lbs)')
plt.ylabel('Frequency')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```
